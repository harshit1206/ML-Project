## Step 4: Train YOLOv8 on Cloud GPU
Using YOLOv8m (Medium) or YOLOv8s (Small) pretrained on COCO.

```python
from ultralytics import YOLO

# Load pretrained YOLO model
model = YOLO('yolov8m.pt')

# Train on GPU
results = model.train(
    data='data.yaml',       # Path to your dataset data.yaml
    epochs=80,             # Recommended: 50-100 epochs on GPU
    imgsz=640,
    batch=16,
    patience=20,
    device=0,              # 0 for NVIDIA GPU
    project='RoadModel',
    name='yolov8m_road_anomalies',
    save=True
)
```
